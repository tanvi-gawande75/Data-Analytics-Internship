## Statistics for Data Analytics and Data Science

## 1. Mean, Median, and Mode
- Mean: The average of all values. It is calculated by adding all values and dividing by the total number of values.
- Median: The middle value when all values are arranged in ascending order. If there are two middle values, take their average.
- Mode: The value that occurs most frequently in a dataset.
### Example: 10, 20, 20, 30, 70
- Mean = 30
- Median = 20
- Mode = 20
### These measures help us understand the central value of a dataset.
## 2. Variance and Standard Deviation
- Variance: Measures how far values are spread from the mean by calculating the average squared deviation.
- Standard deviation: Measures the spread of values around the mean in the original unit of the data.



## 3. Learn Quartiles and Percentiles
## A. Quartiles
Definition: Quartiles are values that divide ordered data into four parts. They help us understand how data is distributed.

## Types of Quartiles:
1. First Quartile (Q1): The 25th percentile. Approximately 25% of observations fall at or below this value.
2. Second Quartile (Q2): The 50th percentile. It is the median of the dataset.
3. Third Quartile (Q3): The 75th percentile. Approximately 75% of observations fall at or below this value.
## Example:

Consider the following marks arranged in ascending order:

10, 20, 30, 40, 50, 60, 70, 80

Using Pandas' default percentile calculation method:
- Q1 = 25
- Q2 = 45
- Q3 = 65
## These values are the cutoff points for the 25th, 50th, and 75th percentiles, respectively.
## B. Percentiles
Definition: Percentiles are values that divide a dataset into 100 parts. They indicate the position of a value relative to the other observations.

## Examples:
- 25th percentile: Approximately 25% of observations are at or below this value.
- 50th percentile: Approximately 50% of observations are at or below this value. It is the median.
- 75th percentile: Approximately 75% of observations are at or below this value.
- 90th percentile: Approximately 90% of observations are at or below this value.
## For example, if the 75th percentile of students' marks is 65, approximately 75% of students scored 65 or below.

## 4. Study Outliers and Distributions
## A. Outliers
Definition: An outlier is a data value that is unusually high or low compared to the other values in a dataset.

## Example: Consider the salaries below:

25000, 27000, 28000, 28000, 32000, 200000

## Here, 200000 may be an outlier because it is much higher than the other salaries.

## Why do we study outliers?
- To identify unusual values in a dataset.
- To detect possible data entry errors.
- To understand how extreme values affect statistical calculations.
- To improve data analysis and decision-making.
## B. Methods to Detect Outliers
1. IQR (Interquartile Range): Detects outliers using the range between the first and third quartiles.
2. Box Plot: A graph that helps identify possible outliers visually.
3. Z-score: Measures how far a value is from the mean in terms of standard deviations.
## C. Distribution
Definition: Distribution describes how data values are spread or arranged in a dataset and how frequently different values occur.

## Types of distributions:
1. Normal Distribution: Data is approximately symmetrical, with most values near the mean.
2. Right-Skewed Distribution: The distribution has a longer tail on the right side. A few unusually high values may pull the mean upward.
3. Left-Skewed Distribution: The distribution has a longer tail on the left side. A few unusually low values may pull the mean downward.

## 6. Compare Mean and Median for a Dataset Containing Outliers
## A. Mean
Definition: Mean is the average of all values in a dataset. It is calculated by dividing the sum of all values by the total number of values.
## B. Median
Definition: Median is the middle value of a dataset after arranging the values in ascending or descending order. If the number of values is even, the median is the average of the two middle values.
## C. Comparison of Mean and Median
The mean is affected by extreme values (outliers), whereas the median is generally less affected by them.

## Example: Consider the following salaries:

25000, 27000, 28000, 28000, 32000, 200000

## Here, ₹2,00,000 is much higher than the other salaries and may be an outlier.

```python
import pandas as pd
import matplotlib.pyplot as plt

# Create a dataset
df = pd.DataFrame({
    "Study_Hours": [1, 2, 3, 4, 5, 6],
    "Marks": [35, 40, 50, 60, 70, 80],
    "Salary": [25000, 27000, 28000, 28000, 32000, 200000]
})

# 1. Mean, Median and Mode
print("Mean:", df["Salary"].mean())
print("Median:", df["Salary"].median())
print("Mode:", df["Salary"].mode().tolist())

# 2. Variance and Standard Deviation
print("Variance:", df["Salary"].var())
print("Standard Deviation:", df["Salary"].std())

# 3. Quartiles and Percentiles
print("Q1:", df["Salary"].quantile(0.25))
print("Q2:", df["Salary"].quantile(0.50))
print("Q3:", df["Salary"].quantile(0.75))
print("90th Percentile:", df["Salary"].quantile(0.90))

# 4. Detect Outliers using IQR
Q1 = df["Salary"].quantile(0.25)
Q3 = df["Salary"].quantile(0.75)
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = df[(df["Salary"] < lower) |
              (df["Salary"] > upper)]

print("Outliers:")
print(outliers["Salary"])

# 5. Study Distribution
plt.hist(df["Salary"], bins=5)
plt.xlabel("Salary")
plt.ylabel("Frequency")
plt.title("Salary Distribution")
plt.show()

# 6. Calculate Correlation
print("Correlation:",
      df["Study_Hours"].corr(df["Marks"]))

# 7. Compare Mean and Median with and without Outlier
salary_without_outlier = df.loc[
    df["Salary"] != 200000, "Salary"
]

print("Mean with outlier:", df["Salary"].mean())
print("Median with outlier:", df["Salary"].median())

print("Mean without outlier:",
      salary_without_outlier.mean())
print("Median without outlier:",
      salary_without_outlier.median())
```

##  Variables

## Types of Variables (based on the values they store)

Variables can store different types of data, such as
- integers
-  floats
-   strings
-   Boolean values.

```python
name = "Tanvi"
age = 22
print(name)
print(age)
```

## Data Types
Definition: Data types define the kind of value a variable stores in Python.

## Types of Data Types
 ## Integer (int)
Definition: An integer is a whole number without a decimal point.

```python
age = 22
print(age)
```

## Float (float)
Definition: A float is a number that contains a decimal point.

```python
price = 99.50
print(price)
```

## String (str)
Definition: A string is a sequence of characters enclosed in single or double quotation marks.

```python
name = "Tanvi"
print(name)
```

## Boolean (bool)
Definition: A Boolean data type represents one of two values: True or False.

```python
is_present = True
print(is_present)
```

## Operators
Definition: Operators are symbols or keywords used to perform operations on values and variables.

## Types of Operators
## Arithmetic Operators
Definition: Arithmetic operators perform mathematical calculations.
## Operator      |	Meaning           |	Example
     +	       Addition	       10 + 5 = 15
     -	       Subtraction	   10 - 5 = 5
     *	       Multiplication   10 * 5 = 50
    /	       Division	            10 / 5 = 2.0
    %	      Modulus (remainder)	10 % 3 = 1
    **	      Exponentiation	    2 ** 3 = 8
    //      	Floor division     	10 // 3 = 3

```python
a = 10
b = 5

print(a + b)
print(a - b)
print(a * b)
print(a / b)
```

## Comparison Operators
Definition: Comparison operators compare two values and return a Boolean result: True or False.
## Operator	   |               Meaning	        |                     Example
      ==	Equal to	5 == 5 → True
      !=	Not equal to	5 != 3 → True
      >	Greater than	8 > 5 → True
      <	Less than	3 < 5 → True
      >=	Greater than or equal to	5 >= 5 → True
      <=	Less than or equal to	3 <= 5 → True

```python
a = 10
b = 5
print(a > b)
print(a == b)
```

## Logical Operators
Definition: Logical operators combine conditions or reverse a condition's result.
## Operator       	|  Meaning	           |               Example
    and	True when both conditions are true	5 > 2 and 8 > 4 → True
     or	True when at least one condition is true	5 < 2 or 8 > 4 → True
     not	Reverses the Boolean result	not True → False

```python
age = 22
print(age >= 18 and age <= 60)
```

## Assignment Operators
Definition: Assignment operators assign values to variables or update their existing values.
## Operator        | 	        Meaning           |      	Example
        =	Assign a value	    x = 10
        +=	Add and assign	    x += 5
       -=	Subtract and assign	x -= 2
       *=	Multiply and assign	x *= 3
       /=	Divide and assign	x /= 2

```python
x = 10
x += 5
print(x)
```

## Type Conversion
Definition: Type conversion is the process of converting a value from one data type to another.

## Types of Type Conversion
## Implicit Type Conversion
Definition: Implicit type conversion occurs automatically when Python converts a value from one data type to another.

```python
a = 10
b = 2.5
result = a + b

print(result)
print(type(result))
```

## Explicit Type Conversion
Definition: Explicit type conversion occurs when the programmer manually converts a value using functions such as int(), float(), and str().

```python
age = "22"
age = int(age)

print(age)
print(type(age))
```

```python
#Calculate Percentage: Percentage represents a value as a fraction of 100.
obtained_marks = 420
total_marks = 500

percentage = (obtained_marks / total_marks) * 100
print("Percentage:", percentage, "%")

#Calculate Total Sales: Total sales is the sum of the sales amounts from all items or transactions.

sales1 = 1000
sales2 = 2000
sales3 = 1500

total_sales = sales1 + sales2 + sales3
print("Total Sales:", total_sales)


#Calculate Discount and Selling Price: A discount is a reduction in the original price of a product. The selling price is the amount paid after deducting the discount.
      
#formula:      {Discount Amount}={Cost Price}*{Discount %}/{100}
               # {Selling Price}={Cost Price}-{Discount Amount}


cost_price = 1000
discount_percent = 20

discount_amount = cost_price * discount_percent / 100
selling_price = cost_price - discount_amount

print("Discount Amount:", discount_amount)
print("Selling Price:", selling_price)


#Calculate Average

marks1 = 80
marks2 = 70
marks3 = 90

average = (marks1 + marks2 + marks3) / 3
print("Average Marks:", average)



```

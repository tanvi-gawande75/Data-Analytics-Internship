```python

import pandas as pd

data = {
    "Employee_ID": [101, 102, 103, 104, 104, 105, 106],
    "Name": ["Tanvi", "Rahul", "Priya", "Amit", "Amit", None, "Neha"],
    "Age": [22, 25, None, 28, 28, 24, 27],
    "Salary": ["25000", "30000", "28000", "35000", "35000", "32000", "invalid"],
    "Department": ["IT", "hr", " IT ", "Sales", "Sales", "HR", "it"],
    "Joining_Date": [
        "2024-01-10", "15/02/2024", "2024-03-20",
        "25/04/2024", "25/04/2024", "2024-05-15", "unknown"
    ]
}

df = pd.DataFrame(data)

print("Original Dataset:")
display(df)
```

```python

df.to_csv("employee_data_original.csv", index=False)

print("Original dataset saved successfully.")
```

## Step 1: Identify missing values
Meaning: Missing values are entries where data is not available. In Pandas, they can be represented by NaN or None.

```python

print("Missing Values in Each Column:")
print(df.isnull().sum())

```

```python
 #Find the rows containing missing values

missing_rows = df[df.isnull().any(axis=1)]

display(missing_rows)
```

```python
#Handle missing values

df_clean = df.copy()

median_age = df_clean["Age"].median()

df_clean["Age"] = df_clean["Age"].fillna(median_age)

print("Dataset After Handling Missing Ages:")
display(df_clean)
```

```python
#After handling the missing age, check the missing-value counts again.
print("Missing Values After Handling:")
print(df_clean.isnull().sum())
```

## Day 7 — Task 2: Detect and Remove Duplicate Records
Meaning: Duplicate records are rows that repeat data already present in a dataset. Duplicate records can lead to incorrect counts and calculations, so we must identify them before deciding whether to remove them.

## Detect duplicate records
Pandas provides duplicated() to identify duplicate rows.

```python
#Detect duplicate records
print("Duplicate Records:")
print(df_clean.duplicated())
print("______________________")
#Count duplicate records
print("Number of Duplicate Rows:", df_clean.duplicated().sum())

```

## Display duplicate rows
Before removing anything, inspect the duplicate records.

```python

duplicates = df_clean[df_clean.duplicated()]

display(duplicates)
```

## Remove duplicate records
Use drop_duplicates() to remove repeated rows.

```python

df_clean = df_clean.drop_duplicates()

print("Dataset After Removing Duplicates:")
display(df_clean)
```

```python
#Verify the result
print("Duplicate Rows After Cleaning:", df_clean.duplicated().sum())
print("Total Rows After Cleaning:", df_clean.shape[0])
```

```python
#If you want to check duplicates using only the employee ID:
print(df_clean["Employee_ID"].duplicated().sum())
```

## Day 7 — Task 3: Correct Data Types
Meaning: Correcting data types means converting a column into the appropriate data type so that calculations, comparisons, and analysis work correctly.

### For example, salary values may be stored as text instead of numbers. We need to convert them to a numeric type before calculating total or average salary.

```python
#Check the current data types

print(df_clean.dtypes)
```

## Convert Salary into a numeric type
Use pd.to_numeric() to convert the salary values into numbers.

##### Because our dataset contains "invalid", we will use errors="coerce" to convert values that cannot be parsed into missing values (NaN).

```python

df_clean["Salary"] = pd.to_numeric(
    df_clean["Salary"],
    errors="coerce"
)

print(df_clean[["Name", "Salary"]])
print("\nUpdated Data Type:")
print(df_clean["Salary"].dtype)
```

```python
#Identify values that could not be converted
#Since "invalid" was converted into NaN, we should identify the affected row.

print(df_clean[df_clean["Salary"].isnull()])


#This displays the employee record whose salary could not be converted.
```

```python
#Verify the data type
print(df_clean.dtypes)

#The Salary column should now have a numeric data type.



```

## Standardize Inconsistent Text Categories
Meaning: Standardizing text categories means making text values consistent by correcting differences in capitalization, extra spaces, or spelling where the correct value is known.

### For example, 
#### "IT", "it", and " IT " represent the same department but are written differently. If we do not standardize them, our analysis may count them as separate categories.

```python
#Identify inconsistent categories :-First, check the unique values in the Department column.
print(df_clean["Department"].unique())
```

```python
#Remove extra spaces :-Use str.strip() to remove spaces at the beginning and end of text values.

df_clean["Department"] = df_clean["Department"].str.strip()
print(df_clean["Department"].unique())

#Now " IT " becomes "IT".
```

```python
#Standardize capitalization :-Use str.upper() to convert all department names to uppercase.
df_clean["Department"] = df_clean["Department"].str.upper()
print(df_clean["Department"].unique())
```

```python
#Verify the cleaned categories :-Check the department names and their counts.
print(df_clean["Department"].value_counts())

#This shows how many records belong to each department.

#Why is value_counts() useful? It helps analysts check category frequencies and confirm that inconsistent spellings are no longer counted separately.
```

```python
#Other useful methods

For this task, you mainly need str.strip() and str.upper(). Two related methods are:

Method	            Purpose	                                               Example
str.lower()    	Converts text to lowercase	                              "IT" → "it"
str.title()	    Capitalizes the first letter of each word           	"human resources" → "Human Resources"
```

## Convert Date Columns into Proper Datetime Values
Meaning: Date conversion means changing date values stored as text into a proper datetime data type so that we can sort dates, calculate time differences, and analyse data by day, month, or year.

```python
#Check the current data

print(df_clean["Joining_Date"])
print("\nCurrent Data Type:")
print(df_clean["Joining_Date"].dtype)
```

```python
#Convert the column to datetime :-Use pd.to_datetime() to convert the dates.

df_clean["Joining_Date"] = pd.to_datetime(
    df_clean["Joining_Date"],
    format="mixed",
    dayfirst=True,
    errors="coerce"
)

print(df_clean[["Name", "Joining_Date"]])
```

```python
#Check the converted data type
print(df_clean["Joining_Date"].dtype)
```

```python
#Identify invalid or missing dates
print(df_clean[df_clean["Joining_Date"].isna()])
```

```python
#Use dates for analysis :-Once converted, you can extract the year and month from the date column.
df_clean["Joining_Year"] = df_clean["Joining_Date"].dt.year
df_clean["Joining_Month"] = df_clean["Joining_Date"].dt.month
print(df_clean[["Name", "Joining_Date", "Joining_Year", "Joining_Month"]])
```

## Compare the Dataset Before and After Cleaning
Meaning: Comparing data before and after cleaning means checking the original dataset against the cleaned dataset to verify which problems were fixed and whether the data is ready for analysis.

```python
#Compare the datasets

print("Original Dataset:")
display(df)

print("Cleaned Dataset:")
display(df_clean)

print("______________________________")

#Compare the number of rows and columns
print("Original Shape:", df.shape)
print("Cleaned Shape:", df_clean.shape)

print("______________________________")

#Compare missing values

comparison = pd.DataFrame({
    "Original Missing Values": df.isnull().sum(),
    "Cleaned Missing Values": df_clean.isnull().sum()
})

print(comparison)

print("______________________________")

#Compare data types
comparison_types = pd.DataFrame({   
    "Original Data Types": df.dtypes.astype(str),  
    "Cleaned Data Types": df_clean.dtypes.astype(str)
})
print(comparison_types)

print("______________________________")

#Check duplicates after cleaning
print("Original Duplicate Rows:", df.duplicated().sum())
print("Cleaned Duplicate Rows:", df_clean.duplicated().sum())

print("______________________________")

```

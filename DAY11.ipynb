## Data Visualization with Matplotlib and Seaborn

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

data = {
    "Month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun"],
    "Sales": [20000, 25000, 22000, 30000, 35000, 40000],
    "Profit": [3000, 4000, 3500, 5000, 6000, 7500],
    "Customers": [100, 120, 110, 150, 170, 200]
}

df = pd.DataFrame(data)
display(df)
```

## 1. Create bar charts and line charts.
## Bar Chart
A bar chart compares values across different categories.

## Line Chart
A line chart shows trends or changes in data over time.

```python
#barchart
plt.bar(df["Month"], df["Sales"])
plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.show()

#linechart
plt.plot(df["Month"], df["Sales"], marker="o", label="Sales")
plt.plot(df["Month"], df["Profit"], marker="o", label="Profit")

plt.title("Monthly Sales and Profit")
plt.xlabel("Month")
plt.ylabel("Amount")
plt.legend()
plt.show()
```

## barchart
## Business Observation:
Sales increased overall from January to June, with the highest sales in June.
## linechart
## Business Observation: 
Both sales and profit show an overall increasing trend across the six months.

## 2. Create histograms and box plots.  
##  Histogram
A histogram shows the frequency distribution of numerical values.
##  Box Plot
A box plot displays the median, quartiles, spread, and potential outliers in numerical data.

```python
#histo
plt.hist(df["Sales"], bins=5, edgecolor="black")
plt.title("Sales Distribution")
plt.xlabel("Sales")
plt.ylabel("Frequency")
plt.show()



#boxplot
sns.boxplot(x=df["Sales"])
plt.title("Sales Box Plot")
plt.xlabel("Sales")
plt.show()
```

## histogram
## Business Observation: 
The histogram shows how frequently sales values fall within different ranges. More observations are needed to determine the overall distribution pattern reliably.
## boxplot
## Business Observation: 
The box plot helps identify the spread of monthly sales and whether any values may be potential outliers.


## 3. Create scatter plots.  
A scatter plot shows the relationship between two numerical variables.

```python
plt.scatter(df["Sales"], df["Profit"])
plt.title("Sales vs Profit")
plt.xlabel("Sales")
plt.ylabel("Profit")
plt.show()
```

## Business Observation: 
Higher sales are generally associated with higher profit in this sample dataset.

## 4. Build a correlation heatmap. 
## 7. Correlation Heatmap
A correlation heatmap uses colors to represent correlation coefficients between numerical variables.

```python
correlation = df[["Sales", "Profit", "Customers"]].corr()

sns.heatmap(correlation, annot=True, cmap="coolwarm", vmin=-1, vmax=1)
plt.title("Correlation Heatmap")
plt.show()
```

## Business Observation: 
The heatmap helps identify the strength and direction of linear relationships among sales, profit, and customers. Correlation does not prove causation.

##  Titles, Labels, Legends, and Scales
These elements make charts easier to understand.
- Title: Describes what the chart represents.
- X-axis label: Describes the horizontal axis.
- Y-axis label: Describes the vertical axis.
- Legend: Identifies different data series.
- Scale: Determines the range and intervals used on an axis.

```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
sales = [20000, 25000, 22000, 30000, 35000, 40000]

plt.figure(figsize=(8, 5))

plt.bar(months, sales, label="Monthly Sales")

plt.title("Monthly Sales Analysis")
plt.xlabel("Month")
plt.ylabel("Sales Amount (₹)")
plt.legend()

plt.yticks(range(0, 45001, 5000))
plt.ylim(0, 45000)

plt.show()
```

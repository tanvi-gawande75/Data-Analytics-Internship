## 1. File Handling
Definition: File handling is used to create, read, write, and manage files using Python.

## Types of File Operations:
1. Read (r): Reads data from a file.
2. Write (w): Writes data into a file.
3. Append (a): Adds new data to the end of a file.

## 2. Error Handling
Definition: Error handling is used to manage errors during program execution so that the program can handle unexpected situations properly.

## Types / Keywords of Error Handling:
1. try: Contains code that may cause an error.
2. except: Handles the error if it occurs.
3. finally: Executes whether an error occurs or not.
4. else: Executes when no error occurs in the try block.
#### Note: 
try, except, finally, and else are keywords used for exception handling, not separate types of errors.

## A. Definition of CSV
- CSV (Comma-Separated Values): A CSV file is a file format used to store tabular data in rows and columns, where values are separated by commas.
- CSV files are commonly used in data analytics to store and exchange datasets.

## B. Reading a CSV File
Definition: Reading a CSV file means retrieving data stored in a CSV file using Python.






```python
#syntax
import csv

with open("filename.csv", "r") as file:
    reader = csv.reader(file)
    for row in reader:
        print(row)
```

## 3. Practice try, except, and finally
## A. Definition of Error Handling
Error handling is a technique used to manage errors during program execution so that the program can respond to unexpected situations properly.
## B. try Statement
Definition: The try block contains code that may cause an error.

```python
#Syntax:
try:    
    # Code that may cause an error
```

## C. except Statement
Definition: The except block handles an error that occurs in the try block.

```python
#Syntax:
except:    
   # Code to handle the error
```

## D. finally Statement
Definition: The finally block always executes whether an error occurs or not.

```python
Syntax:
finally:  
     # Code that always executes
```

```python
#Example Using try, except, and finally
try:    
    number = int(input("Enter a number: ")) 
    print("Number:", number)
except ValueError:   
    print("Invalid input! Please enter a number.")
finally:    
    print("Program execution completed.")

```

## F. Explanation
- try — Executes the code that may cause an error.
- except ValueError — Handles invalid input when it cannot be converted into an integer.
- finally — Executes in both cases, whether an error occurs or not.
- int() — Converts a value into an integer.

## 5. Create a Program to Read a CSV File, Check Its Existence, and Handle Invalid Data

```python
#Create the CSV file
import csv

with open("sales.csv", "w", newline="") as file:
    writer = csv.writer(file)
    writer.writerow(["Product", "Sales"])
    writer.writerow(["Laptop", 50000])
    writer.writerow(["Mobile", 20000])
    writer.writerow(["Tablet", "invalid"])

print("CSV file created successfully")
```

```python
#Read the file and handle errors
import csv
import os

filename = "sales.csv"

try:
    if not os.path.exists(filename):
        raise FileNotFoundError("CSV file not found")

    total = 0

    with open(filename, "r") as file:
        reader = csv.DictReader(file)

        for row in reader:
            try:
                sales = float(row["Sales"])
                total += sales
                print(row["Product"], sales)

            except ValueError:
                print("Invalid sales data:", row)

    print("Total Sales:", total)

except FileNotFoundError as error:
    print(error)

finally:
    print("Processing completed")
```

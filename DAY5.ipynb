## 3. Practice Indexing, Slicing, Filtering, and Reshaping
These NumPy operations help us access, select, filter, and rearrange array data.
## A. Indexing
Definition: Indexing is used to access a specific element from an array using its position.

```python
import numpy as np

arr = np.array([10, 20, 30, 40, 50])

print(arr[0])
print(arr[2])
print(arr[-1])
```

## Explanation:
- arr[0] accesses the first element.
- arr[2] accesses the third element.
- arr[-1] accesses the last element.
### Remember: NumPy indexing starts at 0.

## B. Slicing
Definition: Slicing is used to access a portion of an array.

## Syntax:
array[start:stop:step]


## The stop index is excluded.

```python
import numpy as np

arr = np.array([10, 20, 30, 40, 50])

print(arr[1:4])
print(arr[:3])
print(arr[::2])
```

## Explanation:
- arr[1:4] selects elements from index 1 to 3.
- arr[:3] selects the first three elements.
- arr[::2] selects every second element.
## C. Filtering
Definition: Filtering is used to select elements from an array that satisfy a specified condition.

```python
import numpy as np

arr = np.array([10, 25, 30, 45, 50])

print(arr[arr > 30])


#Explanation: arr > 30 checks which elements are greater than 30. Only the elements satisfying the condition are returned.

#Another example:
print(arr[arr % 2 == 0])
```

## D. Reshaping
Definition: Reshaping is used to change the dimensions or structure of an array without changing its elements.

## Syntax:
array.reshape(rows, columns)

```python
import numpy as np

arr = np.array([1, 2, 3, 4, 5, 6])

new_arr = arr.reshape(2, 3)

print(new_arr)
```

## 4. Calculate Sum, Mean, Minimum, Maximum, and Standard Deviation
## A. Definition
NumPy provides built-in functions to perform statistical and mathematical calculations on numerical data efficiently.


```python
B. NumPy Functions
Function	Meaning
np.sum()	Calculates the sum of all elements.
np.mean()	Calculates the average of all elements.
np.min()	Returns the smallest element.
np.max()	Returns the largest element.
np.std()	Calculates the standard deviation, which measures how spread out the values are.
```

```python
import numpy as np

numbers = np.array([10, 20, 30, 40, 50])

print("Sum:", np.sum(numbers))
print("Mean:", np.mean(numbers))
print("Minimum:", np.min(numbers))
print("Maximum:", np.max(numbers))
print("Standard Deviation:", np.std(numbers))
```

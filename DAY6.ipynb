```python
#Example:
#First, create a sample CSV file:

import pandas as pd

data = {
    "Name": ["Tanvi", "Rahul", "Priya"],
    "Marks": [85, 90, 78]
}

df = pd.DataFrame(data)
df.to_csv("students.csv", index=False)

#Now, load and display the CSV file:
import pandas as pd

df = pd.read_csv("students.csv")
print(df)
```

## Day 6 — Task 3: Practice head(), tail(), info(), and describe()
These Pandas functions help us inspect a dataset and understand its structure, data types, and statistical information.
## 1. head()
Meaning: The head() function displays the first 5 rows of a dataset by default.

## Syntax:
df.head()

```python
import pandas as pd

data = {
    "Name": ["Tanvi", "Rahul", "Priya", "Amit", "Neha", "Riya"],
    "Age": [22, 24, 21, 23, 22, 25],
    "Marks": [85, 90, 78, 88, 92, 80]
}

df = pd.DataFrame(data)

print(df.head())
```

df.tail()

```python
print(df.tail())
```

Meaning: The info() function provides information about a DataFrame, including the number of rows, columns, column names, data types, and non-null values.

## Syntax:
df.info()

```python
df.info()
```

## 4. describe()
Meaning: The describe() function provides statistical summaries of numerical columns.

## Syntax:
df.describe()

```python
print(df.describe())
```

```python
import pandas as pd

data = {
    "Name": ["Tanvi", "Rahul", "Priya", "Amit", "Neha", "Riya"],
    "Age": [22, 24, 21, 23, 22, 25],
    "Marks": [85, 90, 78, 88, 92, 80]
}

df = pd.DataFrame(data)

print("First 5 Rows:")
print(df.head())

print("\nLast 5 Rows:")
print(df.tail())

print("\nDataset Information:")
df.info()

print("\nStatistical Summary:")
print(df.describe())
```

## Day 6 — Task 4: Row Counts, Column Counts, Data Types, and Column Selection
In Data Analytics, these operations help us understand a dataset, check its structure, and select the information required for analysis.
## 1. Row Count and Column Count
Meaning: Row count tells us how many records a dataset contains. Column count tells us how many variables or fields it contains.

```python
import pandas as pd

data = {
    "Name": ["Tanvi", "Rahul", "Priya", "Amit", "Neha"],
    "Age": [22, 24, 21, 23, 22],
    "Marks": [85, 90, 78, 88, 92]
}

df = pd.DataFrame(data)
print(df)
```

```python
#Count rows:
print(df.shape[0])

#Count columns:
print(df.shape[1])

#Get both counts together:
print(df.shape)
```

## 2. Data Types
Meaning: Data types tell us what kind of values a column contains, such as text, integers, decimal numbers, or dates.

Data type	Meaning	                       Example
int64	    Whole numbers	               22, 85
float64	    Decimal numbers	                85.5
object	    Often text or mixed values  	"Tanvi"
bool	    True or False	                  True
datetime64	Date and time values	        2026-10-25

```python
#Check the data type of every column:
print(df.dtypes)
print("___________________")
#Check the data type of one column:
print(df["Marks"].dtype)
```

## 3. Column Selection
Meaning: Column selection means extracting specific columns from a DataFrame.

```python
#A. Select one column
print(df["Name"])
print("--------------------------")
#B. Select multiple columns
print(df[["Name", "Marks"]])
```

## C. Select rows and columns using loc[]
loc[] selects data using row labels and column names.

```python
print(df.loc[0:2, ["Name", "Marks"]])

#This selects rows with labels 0, 1, and 2, along with the Name and Marks columns. The ending label is included.
```

## D. Select rows and columns using iloc[]
iloc[] selects data using integer positions, starting from 0.

```python
print(df.iloc[0:3, [0, 2]])

#This selects the first three rows and the first and third columns. The ending position in 0:3 is excluded.
```

```python
#E. Select rows using a condition
#This is especially useful in Data Analytics because analysts often need to filter records.
print(df[df["Marks"] > 85])
```

## Day 6 — Task 5: Identify Missing Values and Potential Data-Quality Issues
## 1. Identify Missing Values
Meaning: Missing values are entries that have no available value in a dataset. In Pandas, they are commonly represented by NaN or None.

```python
import pandas as pd

data = {
    "Name": ["Tanvi", "Rahul", None, "Amit", "Neha"],
    "Age": [22, None, 21, 23, 22],
    "Marks": [85, 90, None, 88, 92]
}

df = pd.DataFrame(data)
print(df)
```

```python
#A. Check missing values
print(df.isnull())

#Returns True wherever a value is missing and False otherwise.
print("________________________")
#B. Count missing values in each column
print(df.isnull().sum())

print("________________________")

#C. Check total missing values
print(df.isnull().sum().sum())
print("________________________")

#D. Find rows containing missing values
print(df[df.isnull().any(axis=1)])

#This displays all rows that contain at least one missing value.
```

## 2. Identify Potential Data-Quality Issues
Data-quality issues are problems that make data incomplete, incorrect, inconsistent, or unsuitable for analysis.

Issue	                        Meaning	                                    How to check
Missing values	               Some values are absent	                   df.isnull().sum()
Duplicate records	           The same record appears more than once  	   df.duplicated().sum()
Incorrect data types	       A column has an unsuitable type	                df.dtypes
Inconsistent text	           Same category has different spellings
                               or capitalization	                        df["Name"].unique()
Invalid values	               Values are outside the expected range	    df[df["Age"] < 0]
Unexpected categories	       Values do not match the allowed categories	df["Name"].unique()

```python
import pandas as pd

data = {
    "Name": ["Tanvi", "Rahul", "Rahul", "Priya", None],
    "Age": [22, 24, 24, -5, 21],
    "Marks": [85, 90, 90, 78, None]
}

df = pd.DataFrame(data)

print("Missing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())

print("\nData Types:")
print(df.dtypes)

print("\nInvalid Age Records:")
print(df[df["Age"] < 0])
```

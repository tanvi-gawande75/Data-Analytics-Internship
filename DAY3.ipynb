## Conditional Statements
Definition: Conditional statements are used to make decisions in a program based on whether a condition is true or false.

## Types of Conditional Statements

 if Statement

Definition: The if statement executes a block of code when a given condition is true.

```python
marks = 80
if marks >= 40:
    print("Pass")
```

##  if-else Statement

Definition: The if-else statement executes one block when the condition is true and another block when it is false.

```python
marks = 30
if marks >= 40:
    print("Pass")
else:
    print("Fail")
```

## if-elif-else Statement

Definition: The if-elif-else statement checks multiple conditions and executes the block of the first true condition.

```python
marks = 75
if marks >= 80:
    print("Grade A")
elif marks >= 60:
    print("Grade B")
else:
    print("Grade C")
```

##  Loops
Definition: Loops are used to execute a block of code repeatedly.

## Types of Loops

 ## for Loop

Definition: A for loop is used to iterate over items in a sequence, such as a list, string, or range.

```python
for i in range(1, 6):
    print(i)
```

## while Loop

Definition: A while loop repeatedly executes a block of code as long as the given condition is true.

```python
i = 1
while i <= 5:
    print(i)
    i += 1
```

## Collections in Python
Definition: Collections are used to store multiple items in a single variable.

## Types of Collections

## List

Definition: A list is an ordered and changeable collection that allows duplicate values.

```python
fruits = ["Apple", "Banana", "Apple"]
print(fruits)
```

## Tuple

Definition: A tuple is an ordered collection that allows duplicate values but cannot be changed after creation.

```python
colors = ("Red", "Blue", "Red")
print(colors)
```

## Set

Definition: A set is a collection of unique items that does not allow duplicate values.

```python
numbers = {10, 20, 10, 30}
print(numbers)
```

## Dictionary

Definition: A dictionary stores data in key-value pairs, where each key is used to access its corresponding value.

```python
student = {"name": "Amit", "marks": 85}
print(student["name"])
```

## Functions
Definition: A function is a reusable block of code designed to perform a specific task.

## Types of Functions

## Built-in Functions

Definition: Built-in functions are predefined functions provided by Python.

```python
numbers = [10, 20, 30]
print(sum(numbers))
print(max(numbers))
```

## User-defined Functions

Definition: User-defined functions are created by programmers using the def keyword to perform specific tasks.

```python
def greet():
    print("Hello, Student!")

greet()
```

## Function with Parameters

Definition: A function with parameters accepts values as input to perform a task.

```python
def add(a, b):
    print(a + b)

add(10, 20)
```

## Function with Return Value

Definition: A function with a return value sends a result back to the place where the function was called.

```python
def add(a, b):
    return a + b

result = add(10, 20)
print(result)
```

## Solve exercises involving calculating totals, finding maximum values, counting categories, and removing duplicates. 

```python
#Calculating Total
sales = [1000, 2000, 1500]
total = sum(sales)
print("Total Sales:", total)


#Finding Maximum Value
numbers = [25, 60, 15, 90, 45]
print("Maximum Value:", max(numbers))


#Counting Categories
categories = ["Movie", "TV Show", "Movie", "Movie"]

print("Movie Count:", categories.count("Movie"))
print("TV Show Count:", categories.count("TV Show"))


#Removing Duplicates

numbers = [10, 20, 10, 30, 20]
unique_numbers = list(dict.fromkeys(numbers))
print(unique_numbers)
```
